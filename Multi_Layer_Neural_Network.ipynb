### 1. Import Required Libraries

```python
import numpy as np
import pandas as pd
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

### 2. Load and Explore the Dataset

```python
df = pd.read_csv('student_exam_scores.csv')
print("Dataset Shape:", df.shape)
df.head(10)
```

### 3. Min-Max Normalization & Train/Test Split
Feature scaling maps all values into the range $[0, 1]$:
$$x_{norm} = \frac{x - x_{min}}{x_{max} - x_{min}}$$

```python
X = df[['Study_Hours', 'Attendance_Percent', 'Previous_Test_Score']].values
y = df['Final_Exam_Score'].values

X_min, X_max = X.min(axis=0), X.max(axis=0)
y_min, y_max = y.min(), y.max()

X_norm = (X - X_min) / (X_max - X_min)
y_norm = (y - y_min) / (y_max - y_min)

# Train/Test split (15 train / 5 test)
test_indices = [0, 1, 8, 15, 17]
train_indices = [i for i in range(len(df)) if i not in test_indices]

X_train, y_train = X_norm[train_indices], y_norm[train_indices]
X_test, y_test = X_norm[test_indices], y_norm[test_indices]
print(f"Training samples: {len(X_train)}, Testing samples: {len(X_test)}")
```

### 4. Build and Train Multi-Layer Neural Network (MLNN)
Architecture: **Input (3 nodes) $\rightarrow$ Hidden (3 neurons, Sigmoid activation) $\rightarrow$ Output (1 neuron)**

```python
mlp = MLPRegressor(
    hidden_layer_sizes=(3,),
    activation='logistic',
    solver='lbfgs',
    max_iter=3000,
    random_state=42
)
mlp.fit(X_train, y_train)
print("Training completed successfully!")
```

### 5. Learned Weights and Biases

```python
print("Input to Hidden Weights (W1):\n", mlp.coefs_[0])
print("\nHidden Biases (b1):\n", mlp.intercepts_[0])
print("\nHidden to Output Weights (W2):\n", mlp.coefs_[1])
print("\nOutput Bias (b2):\n", mlp.intercepts_[1])
```

### 6. Model Evaluation and Performance Metrics

```python
test_preds_norm = mlp.predict(X_test)
test_preds = test_preds_norm * (y_max - y_min) + y_min
test_actual = y[test_indices]

all_preds_norm = mlp.predict(X_norm)
all_preds = all_preds_norm * (y_max - y_min) + y_min

mae = mean_absolute_error(test_actual, test_preds)
rmse = root_mean_squared_error(test_actual, test_preds)
r2 = r2_score(test_actual, test_preds)

print(f"Test Mean Absolute Error (MAE): {mae:.4f}")
print(f"Test Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"Test R-squared (R²): {r2:.4f}")
```
